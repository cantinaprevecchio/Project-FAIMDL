# Fine-tuning the COCO model on Cityscapes semantic segmentation


```python
import copy
import yaml
import pprint
import os

CITY_CFG_PATH = "/content/project/eomt/configs/dinov2/cityscapes/semantic/eomt_base_640.yaml"
COCO_CFG_PATH = "/content/project/eomt/configs/dinov2/coco/panoptic/eomt_base_640_2x.yaml"
CUSTOM_CFG_PATH = "/content/project/eomt/configs/dinov2/cityscapes/semantic/eomt_base_640_from_coco_ft.yaml"

RUN_DIR = "/content/work/step5_runs/coco_to_cityscapes_semantic_ft"
CKPT_DIR = os.path.join(RUN_DIR, "checkpoints")
WANDB_DIR = os.path.join(RUN_DIR, "wandb")

os.makedirs(RUN_DIR, exist_ok=True)
os.makedirs(CKPT_DIR, exist_ok=True)
os.makedirs(WANDB_DIR, exist_ok=True)

with open(CITY_CFG_PATH, "r") as f:
    city_cfg = yaml.safe_load(f)

with open(COCO_CFG_PATH, "r") as f:
    coco_cfg = yaml.safe_load(f)

ft_cfg = copy.deepcopy(city_cfg)

# Replace only the internal network definition with the COCO-compatible one to keep the task
# as Cityscapes semantic segmentation while allowing us to initialize from the COCO-trained checkpoint
ft_cfg["model"]["init_args"]["network"] = copy.deepcopy(
    coco_cfg["model"]["init_args"]["network"]
)

# Check if the data section already contains an "init_args" dictionary
if "init_args" not in ft_cfg["data"]:
    ft_cfg["data"]["init_args"] = {}

# Optional: use 2 dataloader workers to reduce Colab
ft_cfg["data"]["init_args"]["num_workers"] = 2

# Replace the trainer section with a clean fine-tuning setup.
ft_cfg["trainer"] = {
    "default_root_dir": RUN_DIR,        # where logs/checkpoints are saved
    "devices": 1,                       # use one GPU
    "precision": "16-mixed",            # AMP / mixed precision
    "max_epochs": 3,                    # number of epochs
    "accumulate_grad_batches": 4,       # gradient accumulation
    "enable_checkpointing": True,       # save checkpoints

    # Configure the experiment logger to keep trace of the metrics
    "logger": {
        "class_path": "lightning.pytorch.loggers.WandbLogger",  #wandb.wandbLoggers (nello yaml su github)
        "init_args": {
            "project": "step5-coco-to-cityscapes-ft",
            "name": "offline_ft_run",
            "save_dir": WANDB_DIR,
            "offline": True
        }
    },

    # to save the checkpoints
    "callbacks": [
        {
            "class_path": "lightning.pytorch.callbacks.ModelCheckpoint",
            "init_args": {
                "dirpath": CKPT_DIR,                # where checkpoints are saved
                "filename": "best",                 # best checkpoint name
                "monitor": "metrics/val_iou_all",   # which metric to observe
                "mode": "max",                      # to look for the best (max) metric
                "save_top_k": 1,                    # to save only the best ckpt
                "save_last": True,                  # to save also the last ckpt
                "auto_insert_metric_name": False    # Evita che Lightning aggiunga automaticamente il nome della metrica nel filename?
            }
        }
    ]
}

with open(CUSTOM_CFG_PATH, "w") as f:
    yaml.safe_dump(ft_cfg, f, sort_keys=False)

print("Saved custom fine-tuning config to:")
print(CUSTOM_CFG_PATH)

print("\nData section:")
pprint.pprint(ft_cfg["data"])

print("\nTrainer section:")
pprint.pprint(ft_cfg["trainer"])
```
